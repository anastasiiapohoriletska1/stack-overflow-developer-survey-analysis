```python
import pandas as pd

df = pd.read_csv('survey_results_public.csv', low_memory=False)
schema = pd.read_csv('survey_results_schema.csv')
print(df.shape)
df.head()
```

Завдання 1. Підрахунок загальної кількості респондентів

```python
total_respondents = len(df)
print(f"Загальна кількість респондентів: {total_respondents}")
```

```python
total_respondents = df['ResponseId'].nunique()
print(f"Загальна кількість респондентів: {total_respondents}")
```

Завдання 2. Аналіз повноти відповідей респондентів

```python
questions = set(schema['qname'].unique()) & set(df.columns)
complete_respondents = df[list(questions)].dropna().shape[0]

print(f"Кількість питань для перевірки: {len(questions)}")
print(f"Респондентів, які відповіли на всі питання: {complete_respondents}")
```

Завдання 3. Статистичний аналіз досвіду респондентів

```python
work_exp_stats = pd.DataFrame({
    'Statistic': ['Mean', 'Median', 'Mode'],
    'Value': [
        round(df['WorkExp'].mean(), 2),
        round(df['WorkExp'].median(), 2),
        df['WorkExp'].mode()[0]
    ]
})

print("Міри центральної тенденції:")
work_exp_stats
```

Завдання 4. Аналіз віддаленої роботи

```python
remote_count = (df['RemoteWork'] == 'Remote').sum()
print(f"Респондентів, які працюють віддалено: {remote_count}")
```

Завдання 5. Визначення популярності Python

```python
df['worked_with_python'] = df['LanguageHaveWorkedWith'].str.contains('python', case=False, na=False)

python_count = df['worked_with_python'].sum()
python_percentage = (python_count / len(df)) * 100

print(f"К-сть респондентів, які працюють з Python: {python_count}")
print(f"Відсоток респондентів: {round(python_percentage, 2)}%")
```

Завдання 6. Аналіз шляхів навчання програмуванню

```python
df['learned_with_online_courses'] = df['LearnCode'].str.contains('Online Courses', case=False, na=False)
online_count = df['learned_with_online_courses'].sum()
print(f"К-сть респондентів, які навчалися через онлайн-курси: {online_count}")
```

Завдання 7. Географічний аналіз компенсації Python-розробників

```python
python_devs = df[df['worked_with_python']].dropna(subset=['ConvertedCompYearly', 'Country'])

py_progers_by_country = python_devs.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median']).reset_index()
py_progers_by_country.columns = ['Country', 'Mean Compensation', 'Median Compensation']
py_progers_by_country = py_progers_by_country.sort_values('Median Compensation', ascending=False)

py_progers_by_country
```

Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів

```python
top5_paid = (
    df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]
    .dropna(subset=['ConvertedCompYearly'])
    .sort_values('ConvertedCompYearly', ascending=False)
    .reset_index(drop=True)
    .head(5)
)

top5_paid
```

Завдання 9. Аналіз популярності Python по віковим категоріям 

```python
python_by_age = df.groupby('Age')['worked_with_python'].mean() * 100
python_by_age = python_by_age.round(2).reset_index()
python_by_age.columns = ['Age Category', 'Python Users %']
python_by_age = python_by_age.sort_values('Python Users %', ascending=False)

python_by_age
```

Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників

```python
percentile_75 = df['ConvertedCompYearly'].quantile(0.75)
print(f"75-й перцентиль за компенсацією: {percentile_75}")

high_paid_remote = df[(df['ConvertedCompYearly'] > percentile_75) & (df['RemoteWork'] == 'Remote')]
top_industries = high_paid_remote['Industry'].value_counts().reset_index()
top_industries.columns = ['Industry', 'Count']
top_industries
```
